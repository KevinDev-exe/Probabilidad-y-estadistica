# **II. Limpieza de Datos**

## **1. Gestión de duplicados**

```python
# Encontrar y borrar datos duplicados base clientes
df_clientes_mod = df_clientes.drop_duplicates(subset=['id_persona'], keep='first')
df_clientes_mod.info() 
```

## **2. Categorizando para encontrar datos incoherentes**

```python
df_clientes_mod['Rango_Edad'] = pd.cut(df_clientes_mod['edad'], bins=[-6, 0, 100, 300], labels=["Negativos", "Coherente", "Incoherente"])
edad_rango = df_clientes_mod.groupby('Rango_Edad').size() # Contando edad por rango 
print(edad_rango)
```

## **3. Transformación de datos negativos e incoherentes en vacíos**

```python
df_clientes_mod['edad'] = pd.to_numeric(df_clientes_mod['edad'], errors='coerce') # Convierte edad en formato numero
df_clientes_mod.loc[(df_clientes_mod['edad']<0)|(df_clientes_mod['edad']>100),'edad']=np.nan
```
