# **II. Limpieza de Datos**

## **1. Gestión de duplicados**

```python
# Encontrar y borrar datos duplicados base clientes
df_clientes_mod = df_clientes.drop_duplicates(subset=['id_persona'], keep='first')
df_clientes_mod.info() 
```

## **2. Categorizando para encontrar datos incoherentes**

```python
# Categorizando la edad para encontrar datos incoherentes
df_clientes_mod['Rango_Edad'] = pd.cut(df_clientes_mod['edad'], bins=[-6, 0, 100, 300], labels=["Negativos", "Coherente", "Incoherente"])
edad_rango = df_clientes_mod.groupby('Rango_Edad').size() # Contando edad por rango 
print(edad_rango)

# Categorizando la frecuencia de visita para encontrar datos incoherentes
df_clientes_mod['Rango_Frecuencia'] = pd.cut(df_clientes_mod['frecuencia_visita'], bins=[-6, -1, 10], labels=["Negativos", "Coherente"])
frecuencia_rango = df_clientes_mod.groupby('Rango_Frecuencia').size() # Contando frecuencia por rango
print(frecuencia_rango)
```

## **3. Transformación de datos negativos e incoherentes en vacíos**

```python
df_clientes_mod['edad'] = pd.to_numeric(df_clientes_mod['edad'], errors='coerce') # Convierte edad en formato numero
df_clientes_mod.loc[(df_clientes_mod['edad']<0)|(df_clientes_mod['edad']>100),'edad']=np.nan

df_clientes_mod['frecuencia_visita'] = pd.to_numeric(df_clientes_mod['frecuencia_visita'], errors='coerce') # Convierte frecuencia en formato numero
df_clientes_mod.loc[df_clientes_mod['frecuencia_visita']<0,'frecuencia_visita']=np.nan
```

## **4. Comprobación de arreglos hechos hasta el momento desde estadística base**

```python
display(df_clientes_mod['edad'].describe())
display(df_clientes_mod['frecuencia_visita'].describe())
```

## **5. Análisis de sensibilidad para variable cuantitativa**

```python
promedio_edad = df_clientes_mod.groupby('consume_licor')['edad'].mean().reset_index()
display(promedio_edad)
```

## **6. Reemplazamos datos vacíos por media o moda según variable de análisis**

```python
# Relleno de la edad con la media según el consumo de licor
df_clientes_mod['edad'] = df_clientes_mod['edad'].fillna(df_clientes_mod.groupby('consume_licor')['edad'].transform('mean')).round().astype(int) # La edad debe ser un número entero

# Relleno de la frecuencia de visita con la media según el estrato socioeconómico
df_clientes_mod['frecuencia_visita'] = df_clientes_mod['frecuencia_visita'].fillna(df_clientes_mod.groupby('estrato_socioeconomico')['frecuencia_visita'].transform('mean')).round().astype(int) # La frecuencia debe ser un número entero

# Relleno del promedio de gasto con la media según el estrato socioeconómico
df_clientes_mod['promedio_gasto_comida'] = df_clientes_mod['promedio_gasto_comida'].fillna(df_clientes_mod.groupby('estrato_socioeconomico')['promedio_gasto_comida'].transform('mean')).round(2) # El gasto es dinero, se deja en 2 decimales
```

## **7. Comprobación de datos completos**

```python
df_clientes_mod[['edad', 'frecuencia_visita', 'promedio_gasto_comida']].info()
```

## **8. Análisis de sensibilidad para variable cualitativa**

```python
conteo_ciudad = df_clientes_mod.groupby(['ciudad_residencia','preferencias_alimenticias'])['preferencias_alimenticias'].size().reset_index(name='conteo')
mas_frecuente = conteo_ciudad.loc[conteo_ciudad.groupby('ciudad_residencia')['conteo'].idxmax()]
display(mas_frecuente)
```

## **9. Relleno de datos con NA para correo y eliminación de columnas innecesarias**

```python
df_clientes_mod['correo_electronico'] = df_clientes_mod['correo_electronico'].fillna("NA")

# Se rellenan también los demás datos cualitativos que quedaron vacíos
df_clientes_mod['telefono_contacto'] = df_clientes_mod['telefono_contacto'].fillna("NA")
df_clientes_mod['preferencias_alimenticias'] = df_clientes_mod['preferencias_alimenticias'].fillna("NA")

# Eliminación de las columnas generadas en el análisis inicial que no son necesarias
df_clientes_mod = df_clientes_mod.drop(['Rango_Edad', 'Rango_Frecuencia'], axis=1)
```
